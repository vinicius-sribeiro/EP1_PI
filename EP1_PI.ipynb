### 1. Definição Teórica e Derivadas Manuais
A Série de Taylor para o cosseno hiperbólico em torno de $x_0 = 0$ é dada por:
$$\cosh(x) = \sum_{n=0}^{\infty} \frac{x^{2n}}{(2n)!} = 1 + \frac{x^2}{2!} + \frac{x^4}{4!} + \frac{x^6}{6!} + \dots$$

**Derivadas Manuais:**
* $f(x) = \cosh(x) \implies f(0) = 1$
* $f'(x) = \sinh(x) \implies f'(0) = 0$
* $f''(x) = \cosh(x) \implies f''(0) = 1$
* $f'''(x) = \sinh(x) \implies f'''(0) = 0$
* $f^{(4)}(x) = \cosh(x) \implies f^{(4)}(0) = 1$

**Limite de N:** Adota-se $N \le 10$ como limite padrão, pois o fatorial no denominador $(2n)!$ cresce rapidamente, garantindo convergência com precisão da máquina (`float64`) com poucos termos.

```python
import matplotlib.pyplot as plt
import numpy as np

def grafico_cosh(fig, ax):
    fig.set_size_inches(15, 6)
    ax.set_xlabel('x')
    ax.set_ylabel('cosh(x)')
    ax.grid(True)

x = np.linspace(-3, 3, 100)
y = np.cosh(x)

fig, ax = plt.subplots()
ax.plot(x, y, linewidth=2.0)
grafico_cosh(fig, ax)
plt.show()
```

```python
import math
import numpy as np

def cosh_taylor(x, n = 3):
    soma = 0
    for i in range(0, n):
        fator = 2 * i
        numerador = np.pow(x, fator)
        denominador = math.factorial(fator)
        soma += numerador / denominador
    return soma

print("Numpy cosh(1): ", np.cosh(1))
print("Taylor cosh(1, n=2): ", cosh_taylor(1, 2))
print("Taylor cosh(1, n=3): ", cosh_taylor(1, 3))
print("Taylor cosh(1, n=5): ", cosh_taylor(1, 5))
```

```python
x = np.linspace(-3, 3, 100)
y1 = np.cosh(x)
y2 = cosh_taylor(x, 2)
y3 = cosh_taylor(x, 3)
y4 = cosh_taylor(x, 4)
y5 = cosh_taylor(x, 5)

fig, ax = plt.subplots()
ax.plot(x, y1, linewidth=2.0, color='lightblue', label='np.cosh')
ax.plot(x, y2, linewidth=1.0, color='black', linestyle='dashed', label='N=2')
ax.plot(x, y3, linewidth=1.0, color='red', linestyle='dashed', label='N=3')
ax.plot(x, y4, linewidth=1.0, color='green', linestyle='dashed', label='N=4')
ax.plot(x, y5, linewidth=1.0, color='purple', linestyle='dashed', label='N=5')

grafico_cosh(fig, ax)
ax.legend()
plt.show()
```

```python
def cosh_otimizado(x, n = 3):
    numerador = 1.0
    denominador = 1.0
    soma = 1.0
    for i in range(1, n):
        fator = 2 * i
        numerador = x * x * numerador
        denominador = fator * (fator - 1) * denominador
        soma += numerador / denominador
    return soma

print("Otimizado cosh(1, n=5):", cosh_otimizado(1, 5))
```

```python
import math
import numpy as np

def cosh_otimizado_com_tabela(x, n = 5):
    x_y_table = {
        0: 1.0,
        1: (math.e + 1/math.e) / 2,
        -1: (math.e + 1/math.e) / 2,
        math.log(2): 1.25,
        -math.log(2): 1.25
    }
    for chave, valor in x_y_table.items():
        if math.isclose(chave, x):
            return valor
    numerador = 1.0
    denominador = 1.0
    soma = 1.0
    for i in range(1, n):
        fator = 2 * i
        numerador = x * x * numerador
        denominador = fator * (fator - 1) * denominador
        a_n = numerador / denominador
        soma = soma + a_n
    return soma

print("Cálculo em x=0 (tabela):", cosh_otimizado_com_tabela(0))
print("Cálculo em x=ln(2) (tabela):", cosh_otimizado_com_tabela(math.log(2)))
```

```python
import numpy as np
import timeit
import matplotlib.pyplot as plt

repetições = 100
n_max = 10
enes = np.arange(1, n_max + 1)
medias = np.zeros(n_max)
erros = np.zeros(n_max)
x = np.linspace(-3, 3, 360)

def calcular_tempo(x, n):
    inicio = timeit.default_timer()
    valor = cosh_otimizado_com_tabela(x, n)
    erro = abs(valor - np.cosh(x))
    return timeit.default_timer() - inicio, erro

tempo_vector = np.vectorize(calcular_tempo)

for n in enes:
    tempos, erro = tempo_vector(x, n)
    medias[n-1] = np.mean(tempos) * 1000 * 1000
    erros[n-1] = np.sum(erro) / 360

fig, ax = plt.subplots()
fig.set_size_inches(20, 8)
ax.plot(enes, medias, linewidth=2.0, color="blue")
ax.grid(True)
ax.set_xlabel('Valor de N')
ax.set_ylabel('tempo em µs', color="blue")
ax.tick_params(axis='y', labelcolor="blue")

ax2 = ax.twinx()
ax2.set_ylabel('Erro Médio', color="red")
ax2.plot(enes, erros, color="red")
ax2.tick_params(axis='y', labelcolor="red")

plt.show()
```
